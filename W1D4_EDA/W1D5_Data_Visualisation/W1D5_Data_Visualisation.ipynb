```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = pd.read_csv("cleaned_data.csv")

print(df.head())
print("\nShape:", df.shape)
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset:")
print(df)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Population_Millions"], bins=5)

plt.title("Population Distribution")
plt.xlabel("Population (Millions)")
plt.ylabel("Number of States")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=df, x="State", y="Population_Millions")

plt.title("Population by State")
plt.xlabel("State")
plt.ylabel("Population (Millions)")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Population_Millions",
    y="Literacy_Rate"
)

plt.title("Population vs Literacy Rate")
plt.xlabel("Population (Millions)")
plt.ylabel("Literacy Rate")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, y="Population_Millions")

plt.title("Population Distribution and Outliers")
plt.ylabel("Population (Millions)")

plt.show()
```
