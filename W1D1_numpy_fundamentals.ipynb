```python
import numpy as np
```

```python
array_1d = np.array([1, 2, 3, 4, 5])

array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:")
print(array_1d)
print("Shape:", array_1d.shape)

print("\n2D Array:")
print(array_2d)
print("Shape:", array_2d.shape)

print("\n3D Array:")
print(array_3d)
print("Shape:", array_3d.shape)
```

```python
numbers = np.array([10, 20, 30, 40])

result = numbers + 5

print("Original:", numbers)
print("After adding 5:", result)
```

```python
numbers = np.array([1, 2, 3, 4, 5])

squared = numbers ** 2
doubled = numbers * 2

print("Original:", numbers)
print("Squared:", squared)
print("Doubled:", doubled)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = A @ B

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nA @ B:")
print(result)
```

```python
df = pd.read_csv("student_scores.csv")

print(df)
```

```python
scores = df["Exam_Score"].to_numpy()

mean_score = np.mean(scores)
std_score = np.std(scores)

print("Mean:", mean_score)
print("Standard deviation:", std_score)
```

```python
hours = df["Hours_Studied"].to_numpy()
scores = df["Exam_Score"].to_numpy()

correlation = np.corrcoef(hours, scores)[0, 1]

print("Correlation:", correlation)
```

```python
test1 = np.array([1, 2, 3])
test2 = np.array([10, 20, 30])
test3 = np.array([5, 10, 15, 20])

print("Test 1:", test1 * 2)
print("Test 2:", test2 * 2)
print("Test 3:", test3 * 2)
```

## Key Learnings

- NumPy arrays are useful for efficient numerical operations.
- 1D, 2D, and 3D arrays have different shapes and dimensions.
- Broadcasting allows operations between compatible arrays and scalars.
- Vectorised operations perform calculations without explicit Python loops.
- The @ operator performs matrix multiplication.
- NumPy can calculate mean, standard deviation, and correlation.
