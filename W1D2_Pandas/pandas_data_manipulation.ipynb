```python
data = {
    "State": ["Karnataka", "Tamil Nadu", "Kerala", "Maharashtra", "Gujarat",
              "Rajasthan", "Punjab", "West Bengal"],
    "Region": ["South", "South", "South", "West", "West",
               "North", "North", "East"],
    "Population_Millions": [61.1, 72.1, 33.4, 112.4, 60.4, 68.5, 27.7, 91.3],
    "Literacy_Rate": [75.4, 80.1, 94.0, 82.3, 78.0, 66.1, 75.8, 76.3]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("indian_states.csv", index=False)

print("Dataset created successfully!")
```

```python
# Load the Indian dataset from CSV

df = pd.read_csv("indian_states.csv")

# Print the shape of the DataFrame
print("Shape:", df.shape)

# Print the data types
print("\nData Types:")
print(df.dtypes)

# Print the first 10 rows
print("\nFirst 10 Rows:")
print(df.head(10))
```

```python
# Filter states with literacy rate above 80%
filtered_df = df[df["Literacy_Rate"] > 80]

print("States with literacy rate above 80%:")
print(filtered_df)
```

```python
# Group states by region and calculate average literacy rate
grouped_df = df.groupby("Region")["Literacy_Rate"].mean()

print("Average literacy rate by region:")
print(grouped_df)
```

```python
# Create additional state information for merge operation

state_info = pd.DataFrame({
    "State": ["Karnataka", "Tamil Nadu", "Kerala", "Maharashtra",
              "Gujarat", "Rajasthan", "Punjab", "West Bengal"],
    "Capital": ["Bengaluru", "Chennai", "Thiruvananthapuram", "Mumbai",
                "Gandhinagar", "Jaipur", "Chandigarh", "Kolkata"]
})

print(state_info)
```

```python
# Merge the main dataset with state information using State as the key

merged_df = pd.merge(df, state_info, on="State", how="inner")

print("Merged DataFrame:")
print(merged_df)
```

```python
# Create a pivot table showing average literacy rate by region

pivot_df = pd.pivot_table(
    df,
    values="Literacy_Rate",
    index="Region",
    aggfunc="mean"
)

print("Pivot Table:")
print(pivot_df)
```

```python
# Export the cleaned DataFrame to CSV
df.to_csv("cleaned_data.csv", index=False)

# Export the cleaned DataFrame to Parquet
df.to_parquet("cleaned_data.parquet", index=False)

print("CSV and Parquet files exported successfully!")
```

```python
# Compare CSV and Parquet file sizes

import os

csv_size = os.path.getsize("cleaned_data.csv")
parquet_size = os.path.getsize("cleaned_data.parquet")

print(f"CSV file size: {csv_size} bytes")
print(f"Parquet file size: {parquet_size} bytes")

if csv_size < parquet_size:
    print("CSV file is smaller.")
else:
    print("Parquet file is smaller.")
```
