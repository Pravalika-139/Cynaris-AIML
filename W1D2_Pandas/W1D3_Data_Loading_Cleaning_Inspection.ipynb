# W1D3: Data Loading, Cleaning & Inspection

This notebook demonstrates how to load, inspect, and clean a dataset using Pandas.

## Objectives
- Load structured data
- Inspect rows, columns, and data types
- Identify missing values
- Identify and remove duplicate records
- Fill missing values
- Verify the cleaned dataset

```python
import pandas as pd
import numpy as np
```

```python
# Create a sample dataset
data = {
    "Name": ["Asha", "Rahul", "Priya", "Rahul", "Kiran"],
    "Age": [20, 21, None, 21, 22],
    "Score": [85, 90, 78, 90, 88]
}

df = pd.DataFrame(data)

print("Dataset loaded successfully!")
print(df)
```

```python
# Inspect the dataset

print("First 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\nSummary statistics:")
print(df.describe())
```

```python
# Check for missing values

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())

# Check for duplicate rows
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

print("\nDuplicate rows:")
print(df[df.duplicated()])
```

```python
# Clean the dataset

# Fill missing Age values with the mean age
mean_age = df["Age"].mean()
df["Age"] = df["Age"].fillna(mean_age)

# Remove duplicate rows
df = df.drop_duplicates()

print("Mean age used for missing value:", mean_age)

print("\nCleaned dataset:")
print(df)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())
```

```python
# Final verification

print("FINAL DATASET")
print(df)

print("\nFinal shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Total duplicate rows:", df.duplicated().sum())

# Basic validation tests
assert df.isnull().sum().sum() == 0
assert df.duplicated().sum() == 0

print("\nAll validation tests passed successfully!")
```
